```python
# Step 3: Develop Comprehensive Urban Cooling Strategy
This notebook takes the "Hot Zones" identified in Step 2 and algorithmically assigns mitigation streams based on zone characteristics like Land Use and Green Space Percentage.

**Mitigation Streams:**
* **3A:** Passive Cooling (Infrastructure & Materials)
* **3B:** Active Cooling (Smart & Behavioral)
```

```python
import pandas as pd

# 1. Load the "Hot Zones" identified from Step 2
# In a real workflow, this might be loaded via pd.read_csv() or from a previous notebook cell
hot_zones_data = [
    {"zone_name": "Downtown Commercial Core", "uhi_intensity_C": 3.5, "primary_land_use": "Commercial", "green_space_pct": 5.0},
    {"zone_name": "Industrial Park East", "uhi_intensity_C": 5.2, "primary_land_use": "Industrial", "green_space_pct": 2.0},
    {"zone_name": "Suburban Residential South", "uhi_intensity_C": 3.1, "primary_land_use": "Residential", "green_space_pct": 25.0}
]

df_zones = pd.DataFrame(hot_zones_data)
display(df_zones)
```

```python
### Define Mitigation Logic
The function below evaluates each urban zone and appends specific actions from the flowchart (3A1, 3A2, 3B1, 3B2) into lists based on conditional thresholds.
```

```python
def generate_strategy(zone):
    passive_actions = []
    active_actions = []

    # ---------------------------------------------------------
    # STREAM 3A: PASSIVE COOLING
    # ---------------------------------------------------------

    # 3A1. Green Infrastructure
    if zone['green_space_pct'] > 15.0:
        passive_actions.append("[3A1] Increase Tree Canopy & Restore Water Bodies")
    else:
        passive_actions.append("[3A1] Install Green Roofs & Green Walls")

    # 3A2. Reflective & Permeable Surfaces
    if zone['primary_land_use'] in ["Industrial", "Commercial"]:
        passive_actions.append("[3A2] Apply Cool (High Albedo) Roofs & Cool Pavement")
    elif zone['primary_land_use'] == "Residential":
        passive_actions.append("[3A2] Use Permeable Pavement for residential streets")

    # ---------------------------------------------------------
    # STREAM 3B: ACTIVE COOLING
    # ---------------------------------------------------------

    # 3B1. Smart Growth & Design
    if zone['primary_land_use'] == "Commercial":
        active_actions.append("[3B1] Enhance Airflow (Wind Canyons) & Implement Misting Systems")

    # 3B2. Community & Public Health
    if zone['uhi_intensity_C'] >= 4.0: # Severe Heat Alert Threshold
        active_actions.append("[3B2] Establish Cooling Centers & Issue Health Warnings")
    else:
        active_actions.append("[3B2] Reduce Anthropogenic Heat (Energy Efficiency)")

    return pd.Series({
        'Passive_Cooling_Strategy': " | ".join(passive_actions),
        'Active_Cooling_Strategy': " | ".join(active_actions)
    })
```

```python
### Execute Strategy Generation
Apply the logic to our identified zones to generate the final comprehensive strategy table.
```

```python
# Apply the strategy function to each row in the dataframe
strategy_results = df_zones.apply(generate_strategy, axis=1)

# Combine the original zone data with the generated strategies
df_comprehensive_plan = pd.concat([df_zones[['zone_name', 'uhi_intensity_C']], strategy_results], axis=1)

# Display the final mitigation plan
print("=== FINAL MITIGATION PLAN GENERATED ===")
print("-> Proceeding to Step 4: PROJECT IMPLEMENTATION\n")

# Use pandas styling to make it readable in the notebook
df_comprehensive_plan.style.set_properties(**{'text-align': 'left'}).set_table_styles([dict(selector='th', props=[('text-align', 'left')])])
```
